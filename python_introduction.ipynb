### 1. String Formatting (Interpolation) with Numbers in Python

When you include numbers inside a string using placeholders like `{}` or `%d`,
you're using **string formatting** (also called **string interpolation**).

Even though you're working with numbers (e.g., integers), Python automatically
converts them to strings when inserting them into the sentence.

### Example with integers

```python
a=87
b=34
c=a+b

print("The sum of", a, "and", b, "is", c)
```

> Assigns integer values to a and b.
> Sums values and store it in c.
> Prints the sentence by passing multiple arguments separated by commas.
> Python automatically inserts spaces between these arguments and converts numbers to strings.

```python
print("The result of adding {} and {} is {}.".format(a,b,c))
```

```python
print('The result of adding %d and %d is %d.' % (a, b, c))
```

# Why f-strings Are Preferred Over Comma-Separated Print Arguments

- **Readability:**  
  F-strings allow you to write the entire message as a single, clear string with variables embedded inside. This looks cleaner than mixing commas and multiple arguments.

- **Control Over Formatting:**  
  With f-strings, you can easily format numbers (like limiting decimal places) or add padding and alignment directly inside the string. Comma-separated print arguments don’t offer this.

- **Performance:**  
  F-strings are faster because the expression is evaluated once and combined into one string before printing. Using commas, Python converts each argument separately and inserts spaces automatically.

- **Flexibility:**  
  F-strings support complex expressions inside `{}`, making it easier to compute or format values inline.


```python
print(f'The result of adding {a} and {b} is {c}.')
```

# Python Variables – A Beginner's Guide

## What is a Variable?

>A **variable** is a name that refers to an **OBJECT** that points to a value stored in memory. An object is like a container that holds data so we can use later in out program. Because a **variable** only points to an object, we don’t need to declare a variable’s type before using it. Python is a dynamically typed language, which means you just assign a value to a variable, and Python figures out the type automatically. Objects where data is stored have their own data **TYPE**!
>
> **Why use variables**
>
>* To use stored data
>* To reuse values without rewriting them
>* To make code readable and organized
>* To perfom calculations and store user input
>
>**Data Types**
>
>* **Numeric**
>   * **Integer**: Whole numbers (1,2,3,4...) positive and negative.
>    * **Float** : Number with decimals, real number.
> * **String**: Text.
>* **Boolean**: True or False
>
> * **Data Structures**: Ordered list of values of the same or different type.
>   * **List** 
>   * **Tuple**
>   * **Dictionary**
>   * **Set**

> **Naming Rules**
> 1. Must start with a letter or underscore(_), Name **CAN NOT** start with a number!
> 2. Only letters, numbers, and underscores (_) can be used in **NAMES**
> 3. Spaces can't be used in names.
> 4. Can **NOT** use Python reserved words (if, for, class, etc)
> 5. **Python is case sensitive**, uppercase letters are different from lower case

```python
age = 25        # Python knows this is an int
name = "Luna"   # Python knows this is a string
price = 19.99   # Python knows this is a float
```

> In Python, we can reassign and make our variable point at a different object (overwrite) a variable at any time — and even change its type. Python simply updates the variable to refer to the new object with the new value stored in it. Whether the object itself changes or a new one is created depends on **mutability**:  
> 
> - If the object is **immutable** (`int`, `float`, `str`, `tuple`), assigning a new value creates a **new object in memory**, and the variable is updated to point to it.  
> - If the object is **mutable** (`list`, `dict`, `set`), changing its content **modifies the existing object in place** without creating a new one.  

```python
x = 10
print(x)   # Output: 10

x = 20     # Since int is immutable data type, new object with value 20 is created.
print(x)   # And our variable points to the newly created object with value 20.

x = "hello"  # Our variable points to object with type str (from int to str)
print(x)   # Output: hello
```

```python
x
```

> To summarize Python is:
>
> * **Dynamically typed**: We don't specify type of variables, it happens automatically
> * **Interpreted**: It updates the variable during execution
> * **Memory-managed**: variables are just names (labels) pointing to objects in memory — and you can change what they point to at any time!

We can check which type of object is our variable pointing to we use the function **type()**:

```python
# Integer
x = 10
print(type(x))  # <class 'int'>

# String
name = "Alice"
print(type(name))  # <class 'str'>

# Float
price = 3.14
print(type(price))  # <class 'float'>

# Boolean
is_valid = True
print(type(is_valid))  # <class 'bool'>

# List
fruits = ["apple", "banana", "cherry"]
print(type(fruits))  # <class 'list'>

# Tuple
coordinates = (10.0, 20.0)
print(type(coordinates))  # <class 'tuple'>

# Dictionary
person = {"name": "Alice", "age": 30}
print(type(person))  # <class 'dict'>

# Set
unique_numbers = {1, 2, 3}
print(type(unique_numbers))  # <class 'set'>

# NoneType
nothing = None
print(type(nothing))  # <class 'NoneType'>
```

Or using f strings

```python
x = 10
print(f"The type of x is {type(x)}")

name = "Alice"
print(f"The type of name is {type(name)}")

price = 3.14
print(f"The type of price is {type(price)}")

is_valid = True
print(f"The type of is_valid is {type(is_valid)}")

fruits = ["apple", "banana", "cherry"]
print(f"The type of fruits is {type(fruits)}")

coordinates = (10.0, 20.0)
print(f"The type of coordinates is {type(coordinates)}")

person = {"name": "Alice", "age": 30}
print(f"The type of person is {type(person)}")

unique_numbers = {1, 2, 3}
print(f"The type of unique_numbers is {type(unique_numbers)}")

nothing = None
print(f"The type of nothing is {type(nothing)}")
```

## 🔹 What is a Module?

A **module** is a file containing Python code — such as functions, classes, and variables — that you can reuse in other Python programs. Modules help keep your code **organized**, **modular**, and **maintainable**.

## What is a String?

A **string** is a sequence of characters enclosed in single (`'`) or double (`"`) quotes. Text in Python is treated as a list of characters. Thus, the methodology that applies to the treatment of lists can be applied to text strings.

String Characteristics:

* **Ordered**: You can access characters by index.

### Common String Methods

```python
s = "hello world"

print(s.upper())       # HELLO WORLD
print(s.lower())       # hello world
print(s.capitalize())  # Hello world
print(s.title())       # Hello World
print(s.strip())       # Removes leading/trailing whitespace
print(s.replace("world", "Python"))  # hello Python
print(s.split())       # ['hello', 'world']
print(len(s))          # 11
```

# 3. Conditional Structures (if)

Conditional control structures allow you to execute one part of the code or another depending on the evaluation of one or several boolean conditions of YES or NO (**<code>TRUE</code>** o **<code>FALSE</code>**).

In Python, conditional control structures are defined by the words **<code>if</code>**, **<code>elif</code>** y **<code>else</code>**.

- **<code>if CONDICION:</code>** if the conditional expression is met, the following code block is executed.
- **<code>elif CONDICION:</code>** otherwise, if this conditional expression is met, this other code block is executed.
- **<code>else:</code>** otherwise, this code block is executed without evaluating any condition.


The condition is usually evaluated by the following relational operators <code> <, <=, ==, >=, >, != </code>.

The part of code inside an **<code>if</code>** cannot be empty, but if for some reason an if statement must be created without content, the **<code>pass</code>** statement can be used to prevent an error from occurring.

| Keyword | Meaning                                                      |
| ------- | ------------------------------------------------------------ |
| `if`    | Executes code **only if** a condition is True                |
| `elif`  | Checks **another condition** if the previous one wasn't True |
| `else`  | Executes code **when all previous conditions are False**     |

### 4.1. Iterative Control Structures (for and while)

Iterative control structures, also known as loops, are used in programming to execute a block of code repeatedly until a certain condition is met.

There are two types of iterative control structures:

* For
* While

The Python ``for`` statement iterates over the elements of any sequence (a list or a string for example), in the order they appear in the sequence.

The ``while`` loop allows multiple iterations to be performed based on the result of a logical expression that can result in a ``True`` or ``False`` value.

Both loops support these control keywords:

* **break** – exits the loop completely.

* **continue** – skips the current iteration and goes to the next

<code> For Loop </code> Used when you want to iterate over a sequence (like a list, tuple, string, or range).


```python
for i in range (0,10,1):
    print(f"Looping: {i}")
```

You can also loop over lists or strings:

```python
for letter in "hello":
    print(f"{letter}")
```

<code> While loop </code>: Used when you want to repeat something as long as a condition is true.

```python
count = 0
while count < 10:
    print(f"{count}")
    count += 1
```

```python
count = 0
while count < 100:
    if count ==5:
        break
    print(f"{count}")
    count += 1
```

```python
for i in range(5):
    if i == 2:
        continue
    print(i)
```

| Use                                                    | Loop    |
| ------------------------------------------------------ | ------- |
| You know how many times to loop                        | `for`   |
| You don't know when to stop; it's based on a condition | `while` |


In addition, loops with ``for`` or ``while`` can be combined with conditional ``if`` structures to create more complex structures.

With the ``break`` statement we can stop the loop before it has gone through all elements.

With the ``continue`` statement we can stop the current iteration of the loop (in both ``for`` and ``while``) and continue with the next one.

```python
numbers = [-5, 3,2, -1, 9, 6, 13, -3, 8]
sum1 = 0;

for n in numbers:
    if n>=0:
        sum1 += n;

sum2 = 0

for n in numbers:
    if n<0:
        continue
    sum2 +=n

sum3 = 0

for n in numbers:
    sum3 +=n

print(f"{sum1}, {sum2}, {sum3}")
```

### 4.2. List Comprehensions

comprehensions are a concise and powerful way to create lists in Python. They let you generate a new list by applying an expression to each item in an existing iterable, optionally filtering items with a condition — all in a single line.

new_list = [expression for item in iterable if condition]

* **expression**: What you want to do with each item (e.g., item * 2)

* **item**: The variable representing each element from the iterable

* **iterable**: Any sequence you want to loop over (like a list, range, string)

* **condition** (optional): A filter that selects which items to include

List comprehensions are a way of fitting a ``for`` loop, an ``if`` statement, and an assignment all on a single line.

A list comprehension consists of the following parts:

+ An input sequence
+ A variable representing members of the input sequence
+ An optional expression
+ An output expression that produces elements of the output list from members of the input sequence that satisfy the predicate

```python
numbers = [1,4,-5,10,-7.2,4,15]
squared = []

for number in numbers:
    squared.append(number**2)
print(f"{type(squared)}")
print(f"{numbers}")
print(f"{squared}")

squared_comp = [number**2 for number in numbers]
print(f"{squared_comp}")
```

### Greatest Common denominator, Greatest Common Divisor (GCD), which is the largest number that divides two or more integers without leaving a remainder.

#### Example: greatest common denominator (GCD)

The common denominator of two positive integers $a$ and $b$ is the greatest common divisor between $a$ and $b$ . 


The Euclidean algorithm is an iterative method for computing the greatest common denominator of two integers. In pseudocode it would be:

+ If $a<b$, change $a$ and $b$.
+ Divide $a$ by $b$ and obtain the residue, $r$. 
+ If $r=0$, the value $b$ is GCD of $a$ and $b$.
+ If $r \neq 0$, iterate again replacing $a$ by $b$ and replacing $b$ by $r$.


```python
def gcd_cgpt(a, b):
    # Step 1: Ensure a >= b
    if a < b:
        a, b = b, a
    
    while True:
        r = a % b  # Step 2: remainder
        if r == 0:  # Step 3: if remainder zero, b is GCD
            return b
        else:       # Step 4: update and repeat
            a, b = b, r
```

The while True: means keep looping indefinitely.

The loop only stops when r == 0, which triggers the return statement.

When return happens, the function ends, so the loop stops.



```python
def gcd_cgpt1(x, y):
    while y: # while y=!0 do iteration, when y=0 loop stops
        x, y = y, x % y
    return x # return x when y=0
```

```python
def GCD(a,b):
    r = 1
    while r != 0:
        if a<b:
            c=a
            a=b
            b=c
        r = a%b 
        if r == 0:
            return b
        else:
            a = b
            b = r
```

Create a function to calculate the factorial of an integer.

The factorial of a non-negative integer $n$ is the product of all positive integers less than or equal to $n$.

```python
def get_factorial(number):
    factorial = 1
    while number > 0:
        factorial *= number
        number -= 1
    return factorial
```

Sometimes it is not known in advance how many items need to be sent to a function.

### 7.1 Map

With a "map" we can apply a function to a whole list of inputs. For example if we want to take a list of numbers and apply the square to all its elements we could do:

```python
numbers = [1,2,3,4,5]
squared = []
for number in numbers:
    squared.append(number**2)
print(f"Numbers in list: {numbers}")
print(f"List of squared numbers: {squared}")
```

Using map:

```python
def squared(number):
    return number**2
squared_list = list(map(squared,numbers))
print(f"Numbers in list: {numbers}")
print(f"List of squared numbers: {squared_list}")
```

As you can see ``map`` is a function that accepts a function as an argument:

``map(function to apply, input list)``

map() returns a map object, which is an iterable.

Wrapping it in list() turns that into a list so you can see/use the values directly.



### 7.2 Filter

The ``Filter`` function, in a similar way, returns a list with those elements of an input list that meet a certain code (filters the input list with a certain condition). Let's see as an example, if we want to remove the negative elements of a list and stay only with the positive ones and that they are pairs

| Function                 | What it does                                                                          | Return value                                   |
| ------------------------ | ------------------------------------------------------------------------------------- | ---------------------------------------------- |
| `map(func, iterable)`    | Applies `func` to **every item** in `iterable`                                        | A new iterable with the **transformed** values |
| `filter(func, iterable)` | Applies `func` to every item and **keeps only the ones where `func(item)` is `True`** | A new iterable with the **filtered** values    |


```python
numbers = [2,5,4,123,364,-342,-54,-653,-234,3242]
positive = []
positive_and_even = []

for number in numbers:
    if number > 0:
        positive.append(number)
        if number%2 == 0:
            positive_and_even.append(number)

print(f"Positive numbers in the list: {positive}")
print(f"Positive and Even numbers in the list: {positive_and_even}")
```

```python
def is_positive(number):
    return number > 0

def is_positive_and_even(number):
    return number > 0 and number % 2 == 0

positive = list(filter(is_positive,numbers))
positive_and_even = list(filter(is_positive_and_even,numbers))

print(f"Positive numbers in the list: {positive}")
print(f"Positive and Even numbers in the list: {positive_and_even}")

```

### 7.3 Reduce

The ``Reduce`` function will be very useful when you want to make a calculation on a list and obtain the result. For example if we want to obtain the result of multiplying the numbers of a list we would do:

```python
product = 1
numbers = [1,5,7,9,11,13,15]
for number in numbers:
    product *= number

print(f"{product}")
```

```python
from functools import reduce

def multiply(x,y):
    return x*y

product1 = reduce(multiply,numbers)
print(f"{product1}")
```

What's happening here:
reduce() applies multiply to the first two numbers: 1 * 5 = 5

Then 5 * 7 = 35

Then 35 * 3 = 105, and so on…

Until it has multiplied all the numbers in the list.



We can see that to use ``reduce`` you have to import it from ``functools`` and that it has a syntax identical to the previous cases:

``reduce(function to apply, input list)``
