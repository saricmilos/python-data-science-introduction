# **1. String Formatting (Interpolation) with Numbers in Python**

When you include numbers inside a string using placeholders like `{}` or `%d`,
you're using **string formatting** (also called **string interpolation**).

Even though you're working with numbers (e.g., integers), Python automatically
converts them to strings when inserting them into the sentence.

### Example with integers

````python
a=87
b=34
c=a+b

print("The sum of", a, "and", b, "is", c)
````

> Assigns integer values to a and b.
> Sums values and store it in c.
> Prints the sentence by passing multiple arguments separated by commas.
> Python automatically inserts spaces between these arguments and converts numbers to strings.

````python
print("The result of adding {} and {} is {}.".format(a,b,c))
````

````python
print('The result of adding %d and %d is %d.' % (a, b, c))
````

# Why f-strings Are Preferred Over Comma-Separated Print Arguments

- **Readability:**  
  F-strings allow you to write the entire message as a single, clear string with variables embedded inside. This looks cleaner than mixing commas and multiple arguments.

- **Control Over Formatting:**  
  With f-strings, you can easily format numbers (like limiting decimal places) or add padding and alignment directly inside the string. Comma-separated print arguments don’t offer this.

- **Performance:**  
  F-strings are faster because the expression is evaluated once and combined into one string before printing. Using commas, Python converts each argument separately and inserts spaces automatically.

- **Flexibility:**  
  F-strings support complex expressions inside `{}`, making it easier to compute or format values inline.


````python
print(f'The result of adding {a} and {b} is {c}.')
````

````python
message = f'The result of adding {a} and {b} is {c}.'
print(message)
````

# **2. Python Variables – A Beginner's Guide**

## What is a Variable?

>A **variable** is a name that refers to an **OBJECT** that points to a value stored in memory. An object is like a container that holds data so we can use later in out program. Because a **variable** only points to an object, we don’t need to declare a variable’s type before using it. Python is a dynamically typed language, which means you just assign a value to a variable, and Python figures out the type automatically. Objects where data is stored have their own data **TYPE**!
>
> **Why use variables**
>
>* To use stored data
>* To reuse values without rewriting them
>* To make code readable and organized
>* To perfom calculations and store user input
>
>**Data Types**
>
>* **Numeric**
>   * **Integer**: Whole numbers (1,2,3,4...) positive and negative.
>    * **Float** : Number with decimals, real number.
> * **String**: Text.
>* **Boolean**: True or False
>
> * **Data Structures**: Ordered list of values of the same or different type.
>   * **List** 
>   * **Tuple**
>   * **Dictionary**
>   * **Set**

> **Naming Rules**
> 1. Must start with a letter or underscore(_), Name **CAN NOT** start with a number!
> 2. Only letters, numbers, and underscores (_) can be used in **NAMES**
> 3. Spaces can't be used in names.
> 4. Can **NOT** use Python reserved words (if, for, class, etc)
> 5. **Python is case sensitive**, uppercase letters are different from lower case

````python
age = 25        # Python knows this is an int
name = "Luna"   # Python knows this is a string
price = 19.99   # Python knows this is a float
````

````python
print(f"My name is {name}, I am {age} years old, and my parents still give me ${price} every week.")
````

> In Python, we can reassign and make our variable point at a different object (overwrite) a variable at any time — and even change its type. Python simply updates the variable to refer to the new object with the new value stored in it. Whether the object itself changes or a new one is created depends on **mutability**:  
> 
> - If the object is **immutable** (`int`, `float`, `str`, `tuple`), assigning a new value creates a **new object in memory**, and the variable is updated to point to it.  
> - If the object is **mutable** (`list`, `dict`, `set`), changing its content **modifies the existing object in place** without creating a new one.  

````python
x = 10
print(x)   # Output: 10

x = 20     # Since int is immutable data type, new object with value 20 is created.
print(x)   # And our variable points to the newly created object with value 20.

x = "hello"  # Our variable points to object with type str (from int to str)
print(x)   # Output: hello
````

````python
x
````

> To summarize Python is:
>
> * **Dynamically typed**: We don't specify type of variables, it happens automatically
> * **Interpreted**: It updates the variable during execution
> * **Memory-managed**: variables are just names (labels) pointing to objects in memory — and you can change what they point to at any time!

We can check which type of object is our variable pointing to we use the function **type()**:

````python
# Integer
x = 10
print(type(x))  # <class 'int'>

# String
name = "Alice"
print(type(name))  # <class 'str'>

# Float
price = 3.14
print(type(price))  # <class 'float'>

# Boolean
is_valid = True
print(type(is_valid))  # <class 'bool'>

# List
fruits = ["apple", "banana", "cherry"]
print(type(fruits))  # <class 'list'>

# Tuple
coordinates = (10.0, 20.0)
print(type(coordinates))  # <class 'tuple'>

# Dictionary
person = {"name": "Alice", "age": 30}
print(type(person))  # <class 'dict'>

# Set
unique_numbers = {1, 2, 3}
print(type(unique_numbers))  # <class 'set'>

# NoneType
nothing = None
print(type(nothing))  # <class 'NoneType'>
````

| Operator | Meaning                     | Example               | Result  |
|----------|-----------------------------|-----------------------|---------|
| `in`     | Checks if value is in list/string/etc | `'a' in 'cat'` | `True`  |
| `not in` | Checks if value is NOT in list/string/etc | `'z' not in 'cat'` | `True` |

````python
# Example values
a = 12
b = 5

print("=== Augmented Assignment Operators Examples ===\n")

# 1. Subtract and assign (x -= b)
x = a
x -= b  # Equivalent to: x = x - b
print(f"x -= b  -> x = {x}  # Subtract b from x ({a}-{b})")

# 2. Multiply and assign (x *= b)
x = a
x *= b  # Equivalent to: x = x * b
print(f"x *= b  -> x = {x}  # Multiply x by b ({a}*{b})")

# 3. Integer division and assign (x //= b)
x = a
x //= b  # Equivalent to: x = x // b
print(f"x //= b -> x = {x}  # Divide x by b using floor division ({a}//{b})")

# 4. Add and assign (x += b)
x = a
x += b  # Equivalent to: x = x + b
print(f"x += b  -> x = {x}  # Add b to x ({a}+{b})")

# 5. Modulus and assign (x %= b)
x = a
x %= b  # Equivalent to: x = x % b
print(f"x %= b  -> x = {x}  # Remainder when x is divided by b ({a}%{b})")

# 6. Exponentiation and assign (x **= b)
x = a
x **= b  # Equivalent to: x = x ** b
print(f"x **= b -> x = {x}  # Raise x to the power of b ({a}^{b})")
````

````python
# Example values
a = 10
b = 3

# Logical AND: True if both conditions are True
result_and = a > b and b < 5
print(f"Is a > b AND b < 5? {result_and}")  # True, because 10>3 AND 3<5

# Logical AND with equality
result_and_eq = a == 10 and b == 4
print(f"Is a == 10 AND b == 4? {result_and_eq}")  # False, because b != 4

# Logical OR: True if at least one condition is True
result_or = a == 5 or b == 3
print(f"Is a == 5 OR b == 3? {result_or}")  
# True, because a == 5 is False OR b == 3 is True → True
````

````python
# Example values
x1, y1 = 8, 4
x2, y2 = 2, 9
x3, y3 = 5, 5
x4, y4 = 10, 0

print("=== Logical Operators Examples ===\n")

# --- AND operator examples ---
# True only if both conditions are True
result_and1 = x1 > y1 and y1 < 10
print(f"x1 = {x1}, y1 = {y1} -> x1 > y1 AND y1 < 10? {result_and1}")
# True because 8>4 AND 4<10

result_and2 = x2 > y2 and y2 < 10
print(f"x2 = {x2}, y2 = {y2} -> x2 > y2 AND y2 < 10? {result_and2}")
# False because 2>9 is False, even though 9<10 is True

# --- OR operator examples ---
# True if at least one condition is True
result_or1 = x3 == 5 or y3 == 10
print(f"x3 = {x3}, y3 = {y3} -> x3 == 5 OR y3 == 10? {result_or1}")
# True because x3 == 5 is True

result_or2 = x4 == 0 or y4 > 0
print(f"x4 = {x4}, y4 = {y4} -> x4 == 0 OR y4 > 0? {result_or2}")
# True because y4 > 0 is True (x4 == 0 is False)

result_or3 = x2 > 10 or y2 < 5
print(f"x2 = {x2}, y2 = {y2} -> x2 > 10 OR y2 < 5? {result_or3}")
# False because both conditions are False

# --- NOT operator examples ---
# Negates the boolean value
result_not1 = not (x1 < y1)
print(f"x1 = {x1}, y1 = {y1} -> NOT(x1 < y1)? {result_not1}")
# True because x1 < y1 is False, NOT False = True

result_not2 = not (y2 > 5)
print(f"x2 = {x2}, y2 = {y2} -> NOT(y2 > 5)? {result_not2}")
# False because y2 > 5 is True, NOT True = False
````

# **4. Python Modules**

---

## What is a Module?

A **module** is a file containing Python code — such as functions, classes, and variables — that you can reuse in other Python programs. Modules help keep your code **organized**, **modular**, and **maintainable**.

---

##  Why Use Modules?

-  Organize large programs into smaller, manageable files
-  Reuse code across multiple programs
-  Access built-in tools and third-party libraries

---

## Types of Modules

1. **Built-in Modules**  
   These come with Python.  
   Examples: `math`, `random`, `datetime`, `os`

2. **External Modules**  
   These need to be installed using tools like `pip`.  
   Examples: `numpy`, `pandas`, `requests`

3. **User-defined Modules**  
   Modules you create yourself by saving code in a `.py` file.

---

````python
# ================================
# Example: Importing Modules in Python
# ================================

# Importing an entire module
import math
value1 = 16
print(f"Square root of {value1} using math module: {math.sqrt(value1)}")
# Output: 4.0

# Importing a specific function from a module
from math import sqrt
value2 = 25
print(f"Square root of {value2} using sqrt directly: {sqrt(value2)}")
# Output: 5.0

# Using an alias for a module
import math as m
print(f"Value of pi using math alias 'm': {m.pi}")
# Output: 3.141592653589793
````

### 5.1.1. Basic List Slicing

Slicing allows you to access a **subset of elements** from a list using the syntax:

```python
list_name[start:stop:step]

````python
numbers = [10, 20, 30, 40, 50, 60, 70, 80]
````

````python
# Slice from index 1 to 3 (elements at positions 1,2,3-1=2)
print(f"numbers[1:4] -> {numbers[1:4]}")  # Output: [20, 30, 40]

# Slice from start to index 2
print(f"numbers[:3] -> {numbers[:3]}")    # Output: [10, 20, 30]

# Slice from index 3 to the end
print(f"numbers[3:] -> {numbers[3:]}")    # Output: [40, 50, 60]
````

````python
print(numbers[0:7:2])  # Output: [10, 30, 50, 70] -> every 2nd element
print(numbers[1::3])   # Output: [20, 50, 80] -> start at index 1, every 3rd element
````

### 5.1.6 Functions vs Methods in Python Lists

In Python, there is an important distinction between **functions** and **methods**:

- **Functions** are general-purpose and can often work on many data types.  
- **Methods** are associated with a specific object (like a list) and are called on that object.  
For lists there are also default Python functions and methods that can be useful. For example the following functions:
- len: returns the length of the list.
- list: create list

To use a function, call function_name(variable_name).

We can also use list-specific methods:
- append: add element to the end of the list.
- insert: add an element in a particular position.
- pop: remove the last element of the list.
- remove: remove a particular element.
- sort: sort the list

To use a method, call variable_name.method_name.

````python
my_list = [3, 1, 4]

my_list.append(5)
print(my_list)       # Output: [3, 1, 4, 5]

my_list.insert(1, 2)
print(my_list)       # Output: [3, 2, 1, 4, 5]

my_list.pop()
print(my_list)       # Output: [3, 2, 1, 4]

my_list.remove(2)
print(my_list)       # Output: [3, 1, 4]

my_list.sort()
print(my_list)       # Output: [1, 3, 4]
````

## 5.2. Tuple (`Tuple`)

Tuples are lists **that cannot be modified.**
- Ordered, **unchangeable** (immutable), allows duplicates
- Defined using parentheses `()`

````python
# Example of a tuple
colors = ("red", "green", "blue")

# Accessing elements by index
print(colors[1])  
# Output: green

# Using f-strings to show type and last element
print(f"The type of 'colors' is {type(colors)} and the last element is {colors[-1]}")
# Output: The type of 'colors' is <class 'tuple'> and the last element is blue

# Attempting to modify a tuple element (will raise an error)
try:
    colors[0] = "Purple"
except TypeError as e:
    print(f"Error: {e}")
# Output: Error: 'tuple' object does not support item assignment
````

## 5.3. Dictionary (`dict`)

A **dictionary** in Python is a collection of **key-value pairs** that is:

- **Unordered** (in Python 3.6+ dictionaries maintain insertion order)  
- **Mutable** (you can add, remove, or change elements)  
- **Indexed by keys** instead of numeric indices  

Dictionaries are defined using **curly braces `{}`**:

````python
# Example: Car dictionary
car = {
    "Brand": "Tesla",
    "Model": "Model S",
    "Year": 2022,
    "Electric": True,
    "Specs": {
        "Horsepower": 670,
        "Top_speed_kmh": 250,
        "Range_km": 652
    },
    "Owners": ["Alice", "Bob"]
}

# Accessing values
print(f"Brand: {car['Brand']}")           # Tesla
print(f"Model: {car['Model']}")           # Model S
print(f"Year: {car['Year']}")             # 2022
print(f"Electric: {car['Electric']}")     # True

# Accessing nested dictionary values
print(f"Horsepower: {car['Specs']['Horsepower']} hp")  # 670

# Accessing elements from a list inside the dictionary
print(f"First owner: {car['Owners'][0]}")             # Alice

# Adding a new key-value pair
car["Color"] = "Red"
print(f"Updated car dictionary: {car}")

# Modifying an existing value
car["Year"] = 2023
print(f"Updated year: {car['Year']}")

# Removing a key
del car["Electric"]
print(f"After removing 'Electric': {car}")
````

## 5.4. Set (`set`)

A **set** in Python is a collection that is:

- **Unordered** (elements have no defined order)  
- **Mutable** (you can add or remove elements)  
- **Unique** (no duplicate values allowed)  

Sets are defined using **curly braces `{}`** or the `set()` function:

````python
# Creating a set
fruits = {"apple", "banana", "orange", "apple"}  # duplicate 'apple' will be removed
print(f"Fruits set: {fruits}")
# Output: Fruits set: {'banana', 'orange', 'apple'}

# Creating a set from a list
numbers = set([1, 2, 3, 2, 4, 4])
print(f"Numbers set: {numbers}")
# Output: Numbers set: {1, 2, 3, 4}

# Adding elements
fruits.add("mango")
print(f"After adding mango: {fruits}")

# Removing elements
fruits.remove("banana")  # raises KeyError if not found
print(f"After removing banana: {fruits}")

# Discarding elements (safe if element not present)
fruits.discard("grape")  # does not raise error
print(f"After discarding grape: {fruits}")

# Set operations
set1 = {1, 2, 3}
set2 = {3, 4, 5}
print(f"Union: {set1 | set2}")        # {1, 2, 3, 4, 5}
print(f"Intersection: {set1 & set2}") # {3}
print(f"Difference: {set1 - set2}")   # {1, 2}
````

| Tuple          | ✅ Yes   | ❌ No    | ✅ Yes                 | `()`            |
| Dictionary     | ✅ Yes\* | ✅ Yes   | ❌ Keys must be unique | `{}`            |
| Set            | ❌ No    | ✅ Yes   | ❌ No                  | `{}` or `set()` |


## 5.5. String

A **string** is a sequence of characters enclosed in single (`'`) or double (`"`) quotes. Text in Python is treated as a list of characters. Thus, the methodology that applies to the treatment of lists can be applied to text strings.

String Characteristics:
* **Immutable**: You can't change characters in a string after it's created.

* **Ordered**: You can access characters by index.

* **Iterable**: You can loop through it.

````python
# ================================
# String Examples in Python
# ================================

# Defining strings
text1 = "Hello"
text2 = 'World'
print("Defining strings:")
print(f"text1 = {text1}, text2 = {text2}\n")

# Accessing characters by index
print("Accessing characters by index:")
print(f"First character of text1: {text1[0]}")   # Output: H
print(f"Last character of text2: {text2[-1]}\n") # Output: d

# Iterating through a string
print("Iterating through text1:")
for char in text1:
    print(char)
print()  # empty line for readability

# Attempting to change a character (immutable)
print("Attempting to change a character in a string (immutable):")
try:
    text1[0] = "h"
except TypeError as e:
    print(f"Error: {e}\n")

# Using quotation marks inside strings
print("Using quotation marks inside strings:")
print("Example 1:", "This is 'an example' inside double quotes")
print("Example 2:", 'Here is "another example" inside single quotes\n')

# Concatenation
print("String concatenation:")
a = 'He'
b = 'llo'
c = a + b + '!'
print(f"Concatenated string: {c}\n")  # Output: Hello!
````

````python
# Common string methods
s = "hello world"
print("Common string methods:")
print(f"Uppercase: {s.upper()}")
print(f"Lowercase: {s.lower()}")
print(f"Capitalize: {s.capitalize()}")
print(f"Title: {s.title()}")
print(f"Strip (remove whitespace): {s.strip()}")
print(f"Replace 'world' with 'Python': {s.replace('world', 'Python')}")
print(f"Split into list: {s.split()}")
print(f"Length of string: {len(s)}")
````

````python
# ================================
# String Slicing Examples
# ================================

a = 'Python'

print("Full string (a[:]):")
print(a[:], "\n")  # Output: 'Python'

print("First character (a[0]):")
print(a[0], "\n")  # Output: 'P'

print("From index 2 to end (a[2:]):")
print(a[2:], "\n")  # Output: 'thon'

print("From start to index 2 (a[:3]):")
print(a[:3], "\n")  # Output: 'Pyt'

print("From index 2 to 3 (a[2:4]):")
print(a[2:4], "\n")  # Output: 'th'

print("Every 2nd character (a[::2]):")
print(a[::2], "\n")  # Output: 'Pto'

print("Every 2nd character starting at index 1 (a[1::2]):")
print(a[1::2], "\n")  # Output: 'yhn'

# Another example using 'word'
word = "Python"

print("Slice word[0:3]:")
print(word[0:3], "\n")  # Output: 'Pyt'

print("Slice word[2:]:")
print(word[2:], "\n")  # Output: 'thon'

print("Slice word[:4]:")
print(word[:4], "\n")  # Output: 'Pyth'

print("Reverse the string word[::-1]:")
print(word[::-1], "\n")  # Output: 'nohtyP'
````

# **6. Conditional Control Structures in Python**

Conditional control structures allow you to execute **different parts of code** depending on whether certain **conditions evaluate to `True` or `False`**. This is fundamental for decision-making in programs.

In Python, conditional structures are defined using the keywords:

- `if`
- `elif`
- `else`

### 6.1. Syntax and Explanation

- **`if CONDITION:`**  
  Executes the following block of code **only if the condition is `True`**.

- **`elif CONDITION:`** *(optional)*  
  Executes this block **if the previous `if` or `elif` conditions were `False`** and this condition is `True`.

- **`else:`** *(optional)*  
  Executes this block **if none of the previous conditions were `True`**. No condition is evaluated.

### 6.2. Conditions and Operators

Conditions in Python are usually evaluated using **relational (comparison) operators**:

| Operator | Meaning                     |
|----------|-----------------------------|
| `<`      | Less than                   |
| `<=`     | Less than or equal to       |
| `==`     | Equal to                    |
| `>=`     | Greater than or equal to    |
| `>`      | Greater than                |
| `!=`     | Not equal to                |


| Keyword | Meaning                                                      |
| ------- | ------------------------------------------------------------ |
| `if`    | Executes code **only if** a condition is True                |
| `elif`  | Checks **another condition** if the previous one wasn't True |
| `else`  | Executes code **when all previous conditions are False**     |


````python
# ================================
# Conditional Control Structures Examples
# ================================

print("Example 1: Simple if statement")
age = 18
if age >= 18:
    print("You are an adult.")
# Output: You are an adult.

print("\nExample 2: if-else statement")
age = 15
if age >= 18:
    print("You are an adult.")
else:
    print("You are a minor.")
# Output: You are a minor.

print("\nExample 3: if-elif-else statement")
score = 85
if score >= 90:
    print("Grade: A")
elif score >= 75:
    print("Grade: B")
elif score >= 60:
    print("Grade: C")
else:
    print("Grade: F")
# Output: Grade: B

print("\nExample 4: Using logical operators (and, or, not)")
temperature = 25
raining = False

if temperature > 20 and not raining:
    print("It's a nice day for a walk!")
else:
    print("Better to stay indoors.")
# Output: It's a nice day for a walk!

print("\nExample 5: Nested if statements")
number = 12

if number > 0:
    if number % 2 == 0:
        print(f"{number} is positive and even.")
    else:
        print(f"{number} is positive and odd.")
else:
    print(f"{number} is zero or negative.")
# Output: 12 is positive and even.

print("\nExample 6: Checking multiple conditions using membership")
fruits = ["apple", "banana", "orange"]
fruit = "mango"

if fruit in fruits:
    print(f"{fruit} is in the list.")
else:
    print(f"{fruit} is not in the list.")
# Output: mango is not in the list.

print("\nExample 7: Using pass in an if statement")
value = 10
if value > 100:
    pass  # placeholder; does nothing
else:
    print(f"{value} is not greater than 100.")
# Output: 10 is not greater than 100.
````

## 6.3. Single-Line Conditional Statements (Ternary Expressions)

In Python, conditional structures can also be expressed **in a single line** using a **ternary expression**, which is useful for concise decisions.

### Syntax
```python
value_if_true if condition else value_if_false

````python
# Example 1: Simple inline if-else
age = 18
status = "Adult" if age >= 18 else "Minor"
print(status)  # Output: Adult

# Example 2: Inline if-elif-else (nested ternary)
score = 85
grade = "A" if score >= 90 else "B" if score >= 75 else "C"
print(grade)  # Output: B

# Example 3: Nested ternary (multiple conditions)
number = 12
result = "Positive and even" if number > 0 and number % 2 == 0 else \
         "Positive and odd" if number > 0 else "Zero or negative"
print(result)  # Output: Positive and even
````
